```python
import sys
import platform

print("=== مشخصات سیستم و مفسر ===")
print(f"نسخه پایتون: {sys.version}")
print(f"مسیر اجرایی پایتون: {sys.executable}")
print(f"سیستم‌عامل: {platform.system()} {platform.release()}")
```

```python
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt

print("=== نگارش کتابخانه‌ها ===")
print(f"NumPy Version     : {np.__version__}")
print(f"Pandas Version    : {pd.__version__}")
print(f"Matplotlib Version: {matplotlib.__version__}")

# رسم نمودار تستی برای تایید صحت محیط
x = np.array(["Python 3.11", "NumPy", "Pandas", "Matplotlib"])
y = np.array([1, 1, 1, 1])
plt.figure(figsize=(7, 2.5))
plt.barh(x, y, color="#4f46e5")
plt.title("Environment Validation Passed")
plt.xlim(0, 1.5)
plt.tight_layout()
plt.show()
```
