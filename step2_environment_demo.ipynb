# آزمایشگاه هوش مصنوعی - گام دوم
## نوت‌بوک جامع آزمون‌های اعتبارسنجی محیط بازی دوز (Tic-Tac-Toe Environment Demo)
* **توسعه‌دهنده محیط:** عرفان دهستانی
* **طراح آزمون‌ها و اعتبارسنجی:** محمدامین فراتی
* **گروه:** شماره ۱

```python
import numpy as np
from tictactoe_env import TicTacToeEnv

print("کتابخانه NumPy و کلاس TicTacToeEnv با موفقیت بارگذاری شدند.")
```

```python
print("=== آزمون ۲: بررسی روند تغییر نوبت و خانه‌های مجاز ===")

env = TicTacToeEnv()


next_state, winner, done, _ = env.step(0)
assert env.board[0] == 1, "خطا: خانه ۰ باید مهره ۱ (X) داشته باشد."
assert env.current_player == -1, "خطا: نوبت باید به بازیکن ۱- (O) منتقل شده باشد."
assert 0 not in env.get_legal_actions(), "خطا: خانه ۰ نباید در لیست حرکات مجاز باشد."
assert done == False, "خطا: بازی نباید در حرکت اول تمام شود."


next_state, winner, done, _ = env.step(4)
assert env.board[4] == -1, "خطا: خانه ۴ باید مهره ۱- (O) داشته باشد."
assert env.current_player == 1, "خطا: نوبت باید دوباره به بازیکن ۱ (X) بازگردد."
assert len(env.get_legal_actions()) == 7, "خطا: اکنون باید دقیقاً ۷ خانه خالی مانده باشد."

print(" آزمون ۲ با موفقیت پاس شد: نوبت‌دهی متناوب و ثبت حرکات بدون نقص است.")
```

```python
print("=== آزمون ۳: اعتبارسنجی مدیریت خطاهای ورودی (Exception Handling) ===")
env.reset()
env.step(0)


try:
    env.step(0)
    assert False, "خطا: بازی در خانه پر شده باید ValueError پرتاب کند!"
except ValueError as e:
    print(f" مهار خطای خانه پر: {e}")


try:
    env.step(9)
    assert False, "خطا: اندیس ۹ خارج از محدوده است و باید خطا دهد!"
except ValueError as e:
    print(f" مهار خطای خارج از محدوده: {e}")

try:
    env.step(-1)
    assert False, "خطا: اندیس منفی باید خطا دهد!"
except ValueError as e:
    print(f" مهار خطای اندیس منفی: {e}")


assert env.current_player == -1, "خطا: پس از حرکت ناموفق، نوبت نباید تغییر کرده باشد."
assert len(env.get_legal_actions()) == 8, "خطا: تعداد خانه‌های مجاز نباید تغییر کند."

print(" آزمون ۳ با موفقیت پاس شد: تمام استثناها و ورودی‌های نامعتبر مهار می‌شوند.")
```

```python
print("=== آزمون ۴: ارزیابی ریاضی هر ۸ خط برد برای بازیکنان X و O ===")


win_test_cases = [
    ("سطر اول (0, 1, 2)", [0, 3, 1, 4, 2], 1),       # X می‌برد
    ("سطر دوم (3, 4, 5)", [3, 0, 4, 1, 5], 1),       # X می‌برد
    ("سطر سوم (6, 7, 8)", [6, 0, 7, 1, 8], 1),       # X می‌برد
    ("ستون اول (0, 3, 6)", [0, 1, 3, 2, 6], 1),      # X می‌برد
    ("ستون دوم (1, 4, 7)", [1, 0, 4, 2, 7], 1),      # X می‌برد
    ("ستون سوم (2, 5, 8)", [2, 0, 5, 1, 8], 1),      # X می‌برد
    ("قطر اصلی (0, 4, 8)", [0, 1, 4, 2, 8], 1),      # X می‌برد
    ("قطر فرعی (2, 4, 6)", [2, 0, 4, 1, 6], 1),      # X می‌برد
    ("برد بازیکن O در سطر اول", [3, 0, 4, 1, 8, 2], -1) # O می‌برد
]

for name, moves, expected_winner in win_test_cases:
    env.reset()
    for m in moves:
        _, winner, done, _ = env.step(m)
    
    assert done == True, f"خطا در سناریوی {name}: بازی باید تمام شده باشد."
    assert winner == expected_winner, f"خطا در سناریوی {name}: برنده باید {expected_winner} باشد اما {winner} شد."
    assert env.get_legal_actions() == [], f"خطا در سناریوی {name}: پس از پایان بازی نباید حرکت مجازی وجود داشته باشد."
    print(f" تایید سناریو: {name} | برنده: {'X' if winner == 1 else 'O'}")

print(" آزمون ۴ با موفقیت پاس شد: تابع check_winner هر ۸ خط هندسی را بی‌نقص تشخیص می‌دهد.")
```

```python
print("=== آزمون ۵: بررسی حالت تساوی کامل (Draw) ===")
env.reset()

draw_moves = [0, 1, 2, 4, 3, 5, 7, 6, 8]

for move in draw_moves:
    _, winner, done, _ = env.step(move)

assert done == True, "خطا: پس از ۹ حرکت بازی باید تمام شده باشد."
assert winner == 0, f"خطا: در وضعیت تساوی برنده باید ۰ باشد، اما {winner} ثبت شد."
assert 0 not in env.board, "خطا: تمام خانه‌ها باید پر شده باشند."
assert env.get_legal_actions() == [], "خطا: هیچ حرکت مجازی نباید باقی مانده باشد."

print("نمای تخته در حالت تساوی:")
env.render()
print(" آزمون ۵ با موفقیت پاس شد: تشخیص تساوی و پر شدن تخته بدون نقص است.")
```

```python
print("=== آزمون ۶: بررسی عدم نشت حافظه (کپی بودن آرایه‌ها) ===")
env.reset()
state_copy = env.step(0)[0]

state_copy[0] = 999

assert env.board[0] == 1, "خطا: تغییر متغیر خروجی نباید حافظه داخلی محیط را تغییر دهد! (نیاز به board.copy())"

print(" آزمون ۶ با موفقیت پاس شد: محیط دارای امنیت کپسوله‌سازی است.")
```

```python
print("=== آزمون ۷: دموی شبیه‌سازی یک بازی کامل با نمایش مرحله‌به‌مرحله ===")
env.reset()
rng = np.random.default_rng(seed=42)
step_count = 1

env.render()

while not env.done:
    legal = env.get_legal_actions()
    action = rng.choice(legal)
    player_symbol = "X" if env.current_player == 1 else "O"
    
    print(f"گام {step_count}: بازیکن {player_symbol} خانه {action} را انتخاب کرد.")
    env.step(action)
    env.render()
    step_count += 1

if env.winner == 0:
    print(" نتیجه نهایی: بازی مساوی شد!")
else:
    print(f" نتیجه نهایی: بازیکن {'X' if env.winner == 1 else 'O'} پیروز شد!")
```
